```python
import numpy as np
import pandas as pd
import seaborn as sns
```

# Analisis de datos

## Importar datos

```python
df = pd.read_csv('coches-de-segunda-mano-sample.csv')
df
```

```python
df = df.drop(columns= ['url', 'company', 'photos', 'insert_date', 'dealer', 'price_financed', 'version', 'publish_date']).copy()
df
```

```python
df.info()
```

```python
df.dropna(axis= 0, inplace=True)
```

# Analisis de datos

```python
df.info()
```

```python
df.describe()
```

```python
sns.pairplot(df)
```

```python
sns.heatmap(df.corr(numeric_only=True), annot= True, cmap='coolwarm', vmin=-1)
```

```python
import matplotlib.pyplot as plt

# Boxplot
sns.boxplot(x=df['price'])
plt.show()

# IQR - límites de outliers
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(f'Límite inferior: {lower:.0f}')
print(f'Límite superior: {upper:.0f}')
print(f'Outliers: {((df["price"] < lower) | (df["price"] > upper)).sum()}')
```
